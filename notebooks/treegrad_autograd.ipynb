# TreeGrad with Autograd — Baseline

This notebook demonstrates the original autograd-based implementation of TreeGrad.
It converts LightGBM trees into differentiable decision forests and trains them via `autograd`.

```python
import numpy as np
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, roc_auc_score

# TreeGrad autograd implementation
import treegrad as tgd
```

## Generate Dataset

```python
X, y = make_classification(
    n_samples=1000,
    n_classes=3,
    n_informative=8,
    n_redundant=2,
    random_state=42
)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

print(f"Train: {X_train.shape}, Test: {X_test.shape}")
print(f"Classes: {np.unique(y)}")
```

## Train LightGBM Base Model

```python
import lightgbm as lgb

lgb_model = lgb.LGBMClassifier(
    num_leaves=31,
    max_depth=-1,
    learning_rate=0.1,
    n_estimators=50,
    random_state=42
)
lgb_model.fit(X_train, y_train)

lgb_preds = lgb_model.predict_proba(X_test)
print(f"LightGBM Test Accuracy: {accuracy_score(y_test, lgb_model.predict(X_test)):.4f}")
try:
    print(f"LightGBM Test AUC (ovo): {roc_auc_score(y_test, lgb_preds, multi_class='ovr', average='macro'):.4f}")
except ValueError:
    pass
```

## Train TreeGrad (Autograd)

```python
# Step 1: fit base model
tgd_model = tgd.TGDClassifier(
    num_leaves=31,
    max_depth=-1,
    learning_rate=0.1,
    n_estimators=50,
    autograd_config={
        "num_iters": 200,
        "batch_size": 64,
        "step_size": 0.05,
        "verbose": True
    }
)
tgd_model.fit(X_train, y_train)
```

```python
# Step 2: partial_fit to learn differentiable parameters
tgd_model.partial_fit(X_train, y_train)
```

```python
# Evaluate
tgd_preds = tgd_model.predict_proba(X_test)
print(f"TreeGrad (Autograd) Test Accuracy: {accuracy_score(y_test, tgd_model.predict(X_test)):.4f}")
try:
    print(f"TreeGrad (Autograd) Test AUC (ovo): {roc_auc_score(y_test, tgd_preds, multi_class='ovr', average='macro'):.4f}")
except ValueError:
    pass
```

## Online / Incremental Learning with Autograd

Demonstrate that TreeGrad can continue learning from new data:

```python
# Generate new batch of data
X_new, y_new = make_classification(
    n_samples=200,
    n_classes=3,
    n_informative=8,
    random_state=99
)

# Continue learning from the already-fitted model
tgd_model.partial_fit(X_new, y_new)
```

```python
# Predictions after continued learning
tgd_after_preds = tgd_model.predict_proba(X_test)
print(f"TreeGrad (after incremental) Test Accuracy: {accuracy_score(y_test, tgd_model.predict(X_test)):.4f}")
```

## Summary

| Model | Accuracy | Notes |
|-------|----------|-------|
| LightGBM | ~{accuracy_score(y_test, lgb_model.predict(X_test)):.4f} | Static ensemble |
| TreeGrad (autograd) | ~{accuracy_score(y_test, tgd_model.predict(X_test)):.4f} | Differentiable forest |

Key observations:
- LightGBM is the static baseline
- TreeGrad converts tree splits into learnable parameters via autograd
- `partial_fit` enables online/incremental learning
- The autograd approach is slower but enables gradient-based optimization of tree structure
